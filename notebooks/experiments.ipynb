```python
def get_chunks_fixed_size_with_overlap(text: str, chunk_size: int, overlap_fraction: float) -> List[str]:
    """
    Splits a given text into chunks of a fixed size with a specified overlap fraction between consecutive chunks.

    Parameters:
    - text (str): The input text to be split into chunks.
    - chunk_size (int): The number of words each chunk should contain.
    - overlap_fraction (float): The fraction of the chunk size that should overlap with the adjacent chunk.
      For example, an overlap_fraction of 0.2 means 20% of the chunk size will be used as overlap.

    Returns:
    - List[str]: A list of chunks (each a string) where each chunk might overlap with its adjacent chunk.
    """

    # Split the text into individual words
    text_words = text.split()
    
    # Calculate the number of words to overlap between consecutive chunks
    overlap_int = int(chunk_size * overlap_fraction)
    
    # Initialize a list to store the resulting chunks
    chunks = []
    
    # Iterate over text in steps of chunk_size to create chunks
    for i in range(0, len(text_words), chunk_size):
        # Determine the start and end indices for the current chunk,
        # taking into account the overlap with the previous chunk
        chunk_words = text_words[max(i - overlap_int, 0): i + chunk_size]
        
        # Join the selected words to form a chunk string
        chunk = " ".join(chunk_words)
        
        # Append the chunk to the list of chunks
        chunks.append(chunk)
    
    # Return the list of chunks
    return chunks
```

```python
from typing import List, Tuple

def get_chunks_by_section(
    text: str,
    section_boundaries: List[Tuple[str, int]],
    max_chunk_words: int = 400,
    chunk_size: int = 262,
    overlap_fraction: float = 0.2
) -> List[Tuple[str, str]]:
    """
    Splits a document into chunks that respect its logical structure (chapters,
    annexures, etc.), only falling back to fixed-size sub-chunking for sections
    too large to safely embed as a single chunk.

    Parameters:
    - text (str): The full document text.
    - section_boundaries (List[Tuple[str, int]]): (section_label, start_position)
      pairs marking where each logical section begins, sorted by position ascending.
      Example: [("Chapter 1", 7367), ("Chapter 2", 12241), ("Annexure I", 75297)]
    - max_chunk_words (int): Word-count threshold above which a section gets broken
      into smaller fixed-size chunks instead of staying as one piece.
    - chunk_size (int): Chunk size (in words) used when sub-chunking an oversized section.
    - overlap_fraction (float): Overlap fraction used when sub-chunking an oversized section.

    Returns:
    - List[Tuple[str, str]]: (chunk_text, section_label) pairs — every chunk is
      tagged with the section it came from, which becomes your citation metadata later.
    """

    # Add a sentinel marker so the very last real section has a defined end point
    boundaries_with_end = section_boundaries + [("__END__", len(text))]

    all_chunks = []

    # Walk through sections using each one's start and the NEXT one's start as its end
    for i in range(len(boundaries_with_end) - 1):
        section_label, start = boundaries_with_end[i]
        _, end = boundaries_with_end[i + 1]

        section_text = text[start:end].strip()
        if not section_text:
            continue  # skip empty sections, just in case

        word_count = len(section_text.split())

        if word_count <= max_chunk_words:
            # Small enough to keep whole — preserves full context in one chunk
            all_chunks.append((section_text, section_label))
        else:
            # Too large for one chunk — sub-chunk, but ONLY within this section's
            # own text, so no chunk ever crosses into a different chapter/annexure
            sub_chunks = get_chunks_fixed_size_with_overlap(
                section_text, chunk_size, overlap_fraction
            )
            for sub_chunk in sub_chunks:
                all_chunks.append((sub_chunk, section_label))

    return all_chunks
```

```python
from typing import List, Tuple
import math

from langchain_text_splitters import NLTKTextSplitter


def get_chunks_by_section(
    text: str,
    section_boundaries: List[Tuple[str, int]],
    max_chunk_chars: int = 2400,
    target_chunk_chars: int = 2400,
    overlap_fraction: float = 0.2,
) -> List[Tuple[str, str]]:
    """
    Splits a document into chunks that respect its logical structure (chapters,
    annexures, etc.). A section is kept whole if it's small; a section too large
    to embed as one piece is sub-chunked, but its target chunk size is computed
    from that section's OWN character count — not a single fixed size reused
    everywhere — so large sections split into evenly-sized pieces instead of a
    full chunk plus one small leftover.

    Parameters:
    - text (str): The full document text.
    - section_boundaries (List[Tuple[str, int]]): (section_label, start_position)
      pairs, sorted by position ascending.
    - max_chunk_chars (int): Character-count threshold above which a section
      gets sub-chunked instead of staying whole.
    - target_chunk_chars (int): The rough size (in characters) we're aiming for
      per sub-chunk — used only to decide HOW MANY pieces an oversized section
      should split into; the actual per-chunk size is then derived per section.
    - overlap_fraction (float): Overlap fraction used when sub-chunking.

    Returns:
    - List[Tuple[str, str]]: (chunk_text, section_label) pairs.
    """
    boundaries_with_end = section_boundaries + [("__END__", len(text))]
    all_chunks = []

    for i in range(len(boundaries_with_end) - 1):
        section_label, start = boundaries_with_end[i]
        _, end = boundaries_with_end[i + 1]

        section_text = text[start:end].strip()
        if not section_text:
            continue

        char_count = len(section_text)

        if char_count <= max_chunk_chars:
            all_chunks.append((section_text, section_label))
        else:
            sub_chunks = get_chunks_by_sentence_balanced(
                section_text, char_count, target_chunk_chars, overlap_fraction
            )
            for sub_chunk in sub_chunks:
                all_chunks.append((sub_chunk, section_label))

    return all_chunks



def get_chunks_by_sentence_balanced(
    section_text: str,
    char_count: int,
    target_chunk_chars: int,
    overlap_fraction: float,
) -> List[str]:
    """
    Sub-chunks a single oversized section. Instead of always cutting at a fixed
    chunk size, it first works out how many roughly-equal pieces THIS section
    should become, then derives a per-section chunk size from that — so a
    12,000-character annexure and a 4,000-character annexure each get evenly
    sized pieces instead of both being cut in fixed 2,400-character slabs.
    Sentence boundaries (via NLTK) are respected throughout, so no chunk starts
    or ends mid-sentence.
    """
    num_chunks = max(1, round(char_count / target_chunk_chars))#count number of chunks 
    even_chunk_size = math.ceil(char_count / num_chunks)  #make the chunk count even
    even_overlap = int(even_chunk_size * overlap_fraction)  #count the overlap

    splitter = NLTKTextSplitter(
        chunk_size=even_chunk_size,
        chunk_overlap=even_overlap,
    )
    return splitter.split_text(section_text)
```

```python
import re
from collections import Counter


def strip_repeated_headers(text: str, min_occurrences: int = 3, max_header_len: int = 60) -> str:
    """
    Removes repeated running headers/footers from a section's text (e.g. a
    page header like "144 LEARNINGS FROM OTHER STATES" that shows up on
    every page of that section, or the same thing extracted reversed, like
    "SRETTEL FO MUIDNEPMOC" for "COMPENDIUM OF LETTERS").

    Approach: split into lines, and for each line, strip any leading/
    trailing digits (page numbers glued to the header text) before
    counting how often that normalized line repeats. Real body content is
    essentially never a short line that repeats 3+ times verbatim -- only
    page furniture does that -- so any short line clearing that bar gets
    removed entirely.
    """
    lines = text.split("\n")

    def normalize(line: str) -> str:
        stripped = line.strip()
        stripped = re.sub(r"^[\d\s]+", "", stripped)   # strip leading page number
        stripped = re.sub(r"[\d\s]+$", "", stripped)   # strip trailing page number
        return stripped

    normalized_lines = [normalize(l) for l in lines]

    # Only short lines are candidates -- real sentences are never this short
    # on their own line.
    candidates = [nl for nl in normalized_lines if nl and len(nl) <= max_header_len]
    counts = Counter(candidates)
    headers_to_remove = {norm for norm, count in counts.items() if count >= min_occurrences}

    cleaned_lines = [
        line for line, norm in zip(lines, normalized_lines)
        if norm not in headers_to_remove
    ]
    return "\n".join(cleaned_lines)
```

```python
"""
Document-structure-aware chunking -- ONE function, nothing else to define
or import separately.
"""
from typing import List, Tuple
import math

from langchain_text_splitters import NLTKTextSplitter


def get_chunks_by_section(
    text: str,
    section_boundaries: List[Tuple[str, int]],
    max_chunk_chars: int = 2400,
    target_chunk_chars: int = 2400,
    overlap_fraction: float = 0.2,
) -> List[Tuple[str, str]]:
    """
    Splits a full document into chunks that respect its own logical
    structure (chapters, annexures, etc.), instead of cutting it at
    arbitrary, structure-blind positions.

    WHAT PROBLEM THIS SOLVES
    -------------------------
    A naive fixed-size chunker slices a document every N characters/words,
    regardless of what's actually there -- a short, self-contained chapter
    might get glued to the next chapter's opening lines, and a chunk can
    start or end mid-sentence. This function avoids both problems: it
    already knows where each chapter/annexure starts (from
    `section_boundaries`), so it never lets a chunk cross from one section
    into another, and it never cuts inside a sentence.

    HOW IT WORKS, STEP BY STEP
    ---------------------------
    1. A sentinel entry ("__END__", len(text)) is appended to
       `section_boundaries` so the LAST real section has a defined end
       point (every other section's end is simply "wherever the next
       section starts").
    2. The function walks each consecutive pair of boundaries. For section
       i, its text is `text[start_i : start_(i+1)]` -- i.e. it runs from
       its own start up to the next section's start.
    3. For each section, decide what to do based on its length in
       characters:
         a. SMALL section (<= max_chunk_chars): kept as a single, whole
            chunk. This preserves full context for short chapters instead
            of splitting them for no reason.
         b. OVERSIZED section (> max_chunk_chars): sub-chunked in place,
            right here in this same function (no second function call):
              - First, decide HOW MANY roughly-equal pieces this specific
                section should become: num_chunks = round(this section's
                char_count / target_chunk_chars). This is computed fresh
                for every oversized section, using THAT section's own
                size -- not one fixed size reused everywhere -- so a
                12,000-character annexure and a 4,000-character annexure
                each get evenly-sized pieces instead of a full-size chunk
                plus one small, awkward leftover.
              - Then derive this section's own chunk size from that count:
                even_chunk_size = ceil(char_count / num_chunks), and scale
                the overlap to match: even_overlap = even_chunk_size *
                overlap_fraction.
              - Finally, do the actual cutting with LangChain's
                `NLTKTextSplitter`, which first runs NLTK's sentence
                tokenizer over the section (correctly handling
                abbreviations like "Govt.", "Dist.", "Dr.", "No." so they
                are NOT mistaken for sentence endings), then packs whole
                sentences into pieces close to `even_chunk_size`,
                carrying `even_overlap` characters' worth of the previous
                chunk's ending sentence(s) into the start of the next
                chunk. Because it can only stop at a real sentence
                boundary, the actual number of pieces produced can come
                out slightly higher than `num_chunks` predicted -- that's
                expected, not a bug, and is the direct cost of never
                cutting mid-sentence.
    4. Every resulting chunk -- whether kept whole or produced by
       sub-chunking -- is tagged with the label of the section it came
       from, and collected into one flat list.

    PARAMETERS
    ----------
    text : str
        The full document text, as one string.
    section_boundaries : List[Tuple[str, int]]
        (section_label, start_position) pairs marking where each logical
        section begins in `text`, sorted by position ascending.
        Example: [("Chapter 1", 7367), ("Chapter 2", 12241),
                  ("Annexure I", 75297)]
    max_chunk_chars : int, default 2400
        Character-count threshold. A section at or below this size is
        kept whole; above it, the section is sub-chunked.
    target_chunk_chars : int, default 2400
        The rough size (in characters) we're aiming for per sub-chunk.
        Used ONLY to decide how many pieces an oversized section should
        split into -- the actual size used for each section is then
        derived specifically for that section (see step 3b above).
    overlap_fraction : float, default 0.2
        Fraction of a section's derived chunk size to use as overlap
        between consecutive sub-chunks of that section (0.2 = 20%).

    RETURNS
    -------
    List[Tuple[str, str]]
        (chunk_text, section_label) pairs, in document order. This label
        is what becomes your citation metadata later, when chunks are
        stored in Chroma and retrieved to answer a question.
    """
    # Add a sentinel so the very last real section has a defined end point.
    boundaries_with_end = section_boundaries + [("__END__", len(text))]

    all_chunks = []

    for i in range(len(boundaries_with_end) - 1):
        section_label, start = boundaries_with_end[i]
        _, end = boundaries_with_end[i + 1]

        section_text = text[start:end].strip()
        if not section_text:
            continue

        section_text = strip_repeated_headers(section_text)   # <-- new line
        if not section_text:
            continue

        char_count = len(section_text)
        char_count = len(section_text)

        if char_count <= max_chunk_chars:
            # Small enough -- keep the whole section as one chunk, so its
            # full context stays together.
            all_chunks.append((section_text, section_label))
        else:
            # Too large for one chunk -- sub-chunk it, but ONLY within
            # this section's own text, so no chunk ever crosses into a
            # different chapter/annexure.

            # Step 1: how many roughly-equal pieces should THIS section
            # become, given its own size?
            num_chunks = max(1, round(char_count / target_chunk_chars))

            # Step 2: derive this section's own chunk size and overlap
            # from that count (not a single fixed size reused everywhere).
            even_chunk_size = math.ceil(char_count / num_chunks)
            even_overlap = int(even_chunk_size * overlap_fraction)

            # Step 3: do the actual cutting, respecting sentence
            # boundaries (NLTK correctly handles "Govt.", "Dr.", "No."
            # style abbreviations so they aren't mistaken for sentence
            # endings).
            splitter = NLTKTextSplitter(
                chunk_size=even_chunk_size,
                chunk_overlap=even_overlap,
            )
            sub_chunks = splitter.split_text(section_text)

            for sub_chunk in sub_chunks:
                all_chunks.append((sub_chunk, section_label))

    return all_chunks
```

```python
import re
from collections import Counter


def strip_repeated_headers(text: str, min_occurrences: int = 3, min_len: int = 10, max_len: int = 60) -> str:
    """
    Removes repeated running headers/footers from a section's text, even
    when they're fused directly into the middle of a sentence (no newline
    around them) -- which is how this document's extraction actually
    produces them at page boundaries.

    Headers are detected by pattern, not by being on their own line: a run
    of ALL-CAPS letters/spaces/slashes at least `min_len` characters long,
    optionally with a 1-4 digit page number glued to either side (e.g.
    "20 CAPACITY BUILDING/CAPACITY DEVELOPMENT" or the same thing
    extracted in reverse, "TNEMPOLEVED YTICAPAC/GNIDLIUB YTICAPAC 02").
    Real sentences essentially never contain a 10+ character all-caps run,
    so this is a safe signal.

    Forward and reversed extractions of the SAME header are recognized as
    one repeated header (not two separate ones) by counting each match
    under a canonical key -- whichever of (match, match-reversed) sorts
    first alphabetically -- so their occurrences combine toward the
    min_occurrences threshold, and both forms get removed once it's met.
    """
    pattern = re.compile(
        r"\b\d{0,4}\s*([A-Z][A-Z /]{" + str(min_len - 2) + "," + str(max_len) + r"}[A-Z])\s*\d{0,4}\b"
    )
    matches = pattern.findall(text)

    def canonical(m: str) -> str:
        return min(m, m[::-1])

    counts = Counter(canonical(m) for m in matches)
    headers_to_remove = {canonical(m) for m in matches if counts[canonical(m)] >= min_occurrences}

    cleaned = text
    for canon in headers_to_remove:
        for variant in (canon, canon[::-1]):
            cleaned = re.sub(r"\d{0,4}\s*" + re.escape(variant) + r"\s*\d{0,4}", " ", cleaned)
    return cleaned
```

```python
#test all chunking 
def get_chunks_fixed_size(text: str, chunk_size: int) -> List[str]:
    """
    Splits a given text into chunks of a specified fixed size.

    Args:
        text (str): The input text to be split into chunks.
        chunk_size (int): The maximum number of words per chunk.

    Returns:
        List[str]: A list of text chunks, each containing up to 'chunk_size' words.
    """
    # Split the input text into individual words
    text_words = text.split()
    
    # Initialize a list to hold the chunks of words
    chunks = []
    
    # Iterate over the word indices in steps of 'chunk_size'
    for i in range(0, len(text_words), chunk_size):
        # Select a sublist of words from 'i' to 'i + chunk_size'
        chunk_words = text_words[i: i + chunk_size]
        
        # Join the selected words into a single string with spaces in between
        chunk = " ".join(chunk_words)
        
        # Add the chunk to the list of chunks
        chunks.append(chunk)
    
    # Return the list of word chunks
    return chunks
```

```python
def get_chunks_fixed_size_with_overlap(text: str, chunk_size: int, overlap_fraction: float) -> List[str]:
    """
    Splits a given text into chunks of a fixed size with a specified overlap fraction between consecutive chunks.

    Parameters:
    - text (str): The input text to be split into chunks.
    - chunk_size (int): The number of words each chunk should contain.
    - overlap_fraction (float): The fraction of the chunk size that should overlap with the adjacent chunk.
      For example, an overlap_fraction of 0.2 means 20% of the chunk size will be used as overlap.

    Returns:
    - List[str]: A list of chunks (each a string) where each chunk might overlap with its adjacent chunk.
    """

    # Split the text into individual words
    text_words = text.split()
    
    # Calculate the number of words to overlap between consecutive chunks
    overlap_int = int(chunk_size * overlap_fraction)  #50*.2
    
    # Initialize a list to store the resulting chunks
    chunks = []
    
    # Iterate over text in steps of chunk_size to create chunks
    for i in range(0, len(text_words), chunk_size):
        # Determine the start and end indices for the current chunk,
        # taking into account the overlap with the previous chunk
        chunk_words = text_words[max(i - overlap_int, 0): i + chunk_size]
        
        # Join the selected words to form a chunk string
        chunk = " ".join(chunk_words)
        
        # Append the chunk to the list of chunks
        chunks.append(chunk)
    
    # Return the list of chunks
    return chunks
```

```python
def mixed_chunking(source_text):
    """
    Splits the given source_text into chunks using a mix of fixed-size and variable-size chunking.
    It first splits the text by Asciidoc markers and then processes the chunks to ensure they are 
    of appropriate size. Smaller chunks are merged with the next chunk, and larger chunks can be 
    further split at the middle or specific markers within the chunk.

    Args:
    - source_text (str): The text to be chunked.

    Returns:
    - list: A list of text chunks.
    """

    # Split the text by Asciidoc marker
    chunks = source_text.split("\n==")

    # Chunking logic
    new_chunks = []
    chunk_buffer = ""
    min_length = 25

    for chunk in chunks:
        new_buffer = chunk_buffer + chunk  # Create new buffer
        new_buffer_words = new_buffer.split(" ")  # Split into words
        if len(new_buffer_words) < min_length:  # Check whether buffer length is too small
            chunk_buffer = new_buffer  # Carry over to the next chunk
        else:
            new_chunks.append(new_buffer)  # Add to chunks
            chunk_buffer = ""

    if len(chunk_buffer) > 0:
        new_chunks.append(chunk_buffer)  # Add last chunk, if necessary

    return new_chunks
```
